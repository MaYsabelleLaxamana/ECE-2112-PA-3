## A. POSITIONAL AND LABEL-BASED SLICING

This loads the dataset into a Data Frame, displays its shape, and extracts a subset of rows (6 through 10) and specific columns using iloc and column labels.

```python
import pandas as pd
cars = pd.read_csv('cars.csv') #This loads the CSV file into a Pandas Data Frame.

#This displays the shape and complete list of column names.
print("Shape of cars:", cars.shape) 
print("Column names:", cars.columns.tolist())

cars_6_to_10 = cars.iloc[5:10] #This uses positional slicing to extract rows 6 through 10.
cars_6_to_10[['Model', 'mpg', 'cyl', 'hp', 'gear']] #This displays only the required columns using indexing.
```

### B. MODEL LOOKUP

This section uses Boolean conditions on the 'Model' column to locate specific cars, extracting the full Toyota Corolla record and a targeted subset for the Pontiac Firebird.

```python
#This locates and displays the complete row for the Toyota Corolla.
toyota = cars[cars['Model'] == 'Toyota Corolla']

toyota
```

```python
#This uses .loc to filter the Pontiac Firebird and extract specific columns.
pontiac = cars.loc[cars['Model']=='Pontiac Firebird', ['Model', 'mpg', 'hp', 'wt']]

pontiac
```

### C. MULTI-MODEL SUBSETTING

This applies multiple Boolean conditions to extract a distinct Data Frame containing only the Datsun 710, Lotus Europa, and Ferrari Dino, retaining exactly five specific columns.

```python
#This extracts specific columns for three exact car models using multiple Boolean conditions.
selected_cars = cars.loc[(cars['Model']=='Datsun 710')|(cars['Model']=='Lotus Europa')|(cars['Model']=='Ferrari Dino'), ['Model', 'mpg', 'cyl', 'hp', 'gear']]

#This displays the shape and the resulting Data Frame.
print("Shape of selected cars:", selected_cars.shape)
selected_cars
```
